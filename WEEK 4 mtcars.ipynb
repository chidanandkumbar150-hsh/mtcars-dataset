```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load mtcars dataset directly from GitHub
url = "https://raw.githubusercontent.com/selva86/datasets/master/mtcars.csv"

mtcars = pd.read_csv(url)

# Display first 5 rows
print(mtcars.head())
```

```python
# 1. Histogram for 'mpg'

plt.hist(mtcars['mpg'], bins=10, color='skyblue', edgecolor='black')

plt.title('Frequency Distribution of MPG')
plt.xlabel('Miles per Gallon (mpg)')
plt.ylabel('Frequency')

plt.show()
```

```python
# 2. Scatter plot between weight (wt) and mpg

sns.scatterplot(x='wt', y='mpg', data=mtcars)
plt.title('Relationship Between Weight and MPG')
plt.show()
```

```python
# 3. Bar plot for transmission type frequency

mtcars['am'] = mtcars['am'].map({
    0: 'Automatic',
    1: 'Manual'
})

sns.countplot(x='am', data=mtcars)

plt.title('Transmission Type Distribution')
plt.show()
```

```python
# 4. Boxplot of mpg

sns.boxplot(x=mtcars['mpg'])
plt.title('Box Plot of MPG')
plt.show()
```

```python
# Print five-number summary
print("Five-number summary of mpg:")
print(mtcars['mpg'].describe())
```
